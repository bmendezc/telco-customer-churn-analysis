```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import SMOTE

path_dir= r"C:\Users\brisa\telco-customer-churn-analysis\data\processed"

X_train= pd.read_csv(f"{path_dir}\\X_train.csv")
X_test = pd.read_csv(f"{path_dir}\\X_test.csv")  ##load in processed data
y_train =pd.read_csv(f"{path_dir}\\y_train.csv")
y_test= pd.read_csv(f"{path_dir}\\y_test.csv")


numCols = ['tenure', 'MonthlyCharges', 'TotalCharges'] ##identify numeric cols for scaling
scaler = StandardScaler()
##apply z-score normalization on selected high val cols
X_train[numCols] = scaler.fit_transform(X_train[numCols])

X_test[numCols] = scaler.transform(X_test[numCols])

smote =SMOTE(random_state=42) ##aply smote to fix the class imbalance
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)

print(f"the new balance afterr SMOTE is \n{y_train_resampled.value_counts()}")


X_train_resampled.to_csv(f"{path_dir}\\X_train_final.csv", index=False)
y_train_resampled.to_csv(f"{path_dir}\\y_train_final.csv", index=False)
X_test.to_csv(f"{path_dir}\\X_test_final.csv", index=False)
y_test.to_csv(f"{path_dir}\\y_test_final.csv", index=False)

```

Now the classes are perfectly 50/50 balanced.
